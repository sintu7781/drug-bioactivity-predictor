```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "../data/processed/egfr_ic50_curated.csv"
)

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df["activity_class"].value_counts()
```

```python
counts = df["activity_class"].value_counts()

plt.figure(figsize=(6, 4))
counts.plot(kind="bar")
plt.title("EGFR Bioactivity Class Distribution")
plt.xlabel("Activity")
plt.ylabel("Number of Compounds")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["standard_value"],
    bins=50,
)

plt.xlabel("IC50 (nM)")
plt.ylabel("Frequency")
plt.title("EGFR IC50 Distribution")

plt.tight_layout()
plt.show()
```

```python
df["pIC50"] = 9 - np.log10(
    df["standard_value"]
)

plt.figure(figsize=(8, 5))

plt.hist(
    df["pIC50"],
    bins=50,
)

plt.xlabel("pIC50")
plt.ylabel("Frequency")
plt.title("EGFR pIC50 Distribution")

plt.tight_layout()
plt.show()
```
